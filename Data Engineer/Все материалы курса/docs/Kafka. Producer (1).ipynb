```python
from datetime import datetime
import random
import json
import socket

from confluent_kafka import Producer
```

### Создание Producer

Нужно указать топик и сервер, на котором работает брокер Kafka

```python
# CODE HERE
topic = 'quickstart'

config = {'bootstrap.servers': 'localhost:9092', 
          'client.id': socket.gethostname()}

producer = Producer(config)
```

### Генерация сообщений

```python
# CODE HERE

for _ in range(5):
    client_id = random.randint(1, 9000000)
    trade_id = random.randint(1, 9000000)
    order_sum = random.randint(1, 9000000)
    message_id = '_'.join([str(client_id), str(trade_id)])
    
    message = dict()
    message['client_id'] = client_id
    message['trade_id'] = trade_id
    message['order_sum'] = order_sum
    
    message_bytes = str.encode(json.dumps(message))
    print(message_bytes)
    
    producer.produce(
        topic=topic,         # имя топика
        value=message_bytes, # сообщение
        partition=0,         # партиция для записи, иначе будет использован встроенный partitioner
    )
    
producer.flush()   # ждем доставки всех сообщений, что в очереди
```
