```python
def ln_sum(n: int):
    """
    Функция  возвращает  кортеж  из значения и 
    выражения суммы логарифмов чисел от 1 до n
    """
    expr = UnevaluatedExpr(log(1))
    for i in range(2, n + 1):
        expr += UnevaluatedExpr(log(i))
    return expr.simplify().evalf()
```

```python
def product_ln(n: int):
    """
    Функция возвращает кортеж из значения и выра-
    жения логарифма произведения  чисел от 1 до n
    """
    expr = UnevaluatedExpr(1)
    for i in range(2, n + 1):
        expr *= UnevaluatedExpr(i)
    expr = log(expr)
    return expr.simplify().evalf()
```

```python
def transpose(matrix):
    transpose_matrix = np.ndarray(matrix.shape[::-1], matrix.dtype)
    for i in range(matrix.shape[0]):
        for j in range(matrix.shape[1]):
            transpose_matrix[j,i] = matrix[i, j]
    return transpose_matrix
```

```python
a1, a2 = symbols('a1 a2')
MSE = 1 / 4 * ((a1 * 1 + a2 * 2 - 5)**2 + \
               (a1 * 5 + a2 * 3 - 6)**2 + \
               (a1 * 2 + a2 * 4 - 10)**2 + \
               (a1 * 3 + a2 * 7 - 8)**2)
print('MSE =', MSE)

grad_MSE1 = diff(MSE, a1)
grad_MSE2 = diff(MSE, a2)
print(f"grad_MSE: ({grad_MSE1} {grad_MSE2})")

# 1. Начинаем с нулевых коэффициентов (это правильно)
a1 = 0.0
a2 = 0.0

# 2. Шаг обучения
step = 0.01

# 3. Цикл градиентного спуска (вместо ручных шагов)
#    Правило: a ← a - step * градиент 
while True:
    # Считаем градиент в текущей точке
    grad1 = 19.5 * a1 + 23.0 * a2 - 39.5
    grad2 = 23.0 * a1 + 39.0 * a2 - 62.0

    # Делаем шаг (это главное исправление)
    a1 = a1 - step * grad1       
    a2 = a2 - step * grad2   

    # Считаем текущий MSE
    current_MSE = 1 / 4 * ((a1 * 1 + a2 * 2 - 5)**2 + \
                           (a1 * 5 + a2 * 3 - 6)**2 + \
                           (a1 * 2 + a2 * 4 - 10)**2 + \
                           (a1 * 3 + a2 * 7 - 8)**2)

    print(f"a1 = {a1:.5f}, a2 = {a2:.5f}, MSE = {current_MSE:.5f}")

    # Останавливаемся, когда MSE стало достаточно близко к минимуму (~6.24)
    if current_MSE <= 6.25:         # ← условие остановки
        break

print("\nИтоговые коэффициенты:")
print(f"a1 = {a1:.5f}")
print(f"a2 = {a2:.5f}")
print(f"MSE = {current_MSE:.5f}")
```
