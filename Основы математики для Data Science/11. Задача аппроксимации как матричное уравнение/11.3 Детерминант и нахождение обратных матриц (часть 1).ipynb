```python
A = Matrix([[3,4],[10,2]])
B = Matrix([[4,8],[2,4]])
A*B
```

```python
print(Matrix([[3,4],[10,2]]) * Matrix([[4,8],[2,4]]))
# Matrix([[20, 40], [44, 88]])
```

```python
print(Matrix([[4,8],[2,4]]) * Matrix([[3,-4],[10,-2]]))
# Matrix([[92, -32], [46, -16]])
```

```python
print(Matrix([[3,-4,1],[10,-2,2]]) * Matrix([[7,13],[-4,5],[5,6]]))
# Matrix([[42, 25], [88, 132]])
```

```python
print(Matrix([[7,13],[-4,5],[9,3]]) * Matrix([[10,4,5,7],[5,14,6,2]]))
# Matrix([[135, 210, 113, 75], [-15, 54, 10, -18], [105, 78, 63, 69]])
```
