```python
from sympy import *
```

```python
A = Matrix([[2,3,-1],[0,1,1],[-4,-6,2]])
print(A.cofactor_matrix())
# Matrix([[8, -4, 4], [0, 0, 0], [4, -2, 2]])

B = Matrix([[1,0,2],[2,-1,1],[1,3,-1]])
print(B.cofactor_matrix())
# Matrix([[-2, 3, 7], [6, -3, -3], [2, 3, -1]])
```

```python
C = Matrix([[1,3,-5],[0,1,2],[0,0,1]])
print(C.det())          # 1
print(C.inv())
# Matrix([[1, -3, 11], [0, 1, -2], [0, 0, 1]])
```
