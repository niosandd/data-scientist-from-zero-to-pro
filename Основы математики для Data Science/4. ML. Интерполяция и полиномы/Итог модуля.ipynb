```python
import sympy as sp
from sympy.plotting import plot
sp.init_printing(use_unicode=False, wrap_line=False, no_global=True)
```

```python
a4, a3, a2, a1, a0 = sp.symbols('a4, a3, a2, a1, a0')
```

```python
# Степень полинома равна сумме точек - 1 = 5 - 1 = 4 - > полином = а4**4, а3**3 ... a0

# f(-2) = 2 => a4*(-2)**4 + a3*(-2)**3 + a2*(-2)**2 + a1*(-2) + a0 - 2 = 0
# f(-1) = 1 => a4*(-1)**4 + a3*(-1)**3 + a2*(-1)**2 + a1*(-1) + a0 - 1 = 0
# f(0) = 0 => a4*(0)**4 + a3*(0)**3 + a2*(0)**2 + a1*(0) + a0 - 0 = 0
# f(1) = 1 => a4*(1)**4 + a3*(1)**3 + a2*(1)**2 + a1*(1) + a0 - 1 = 0
# f(2) = 2 => a4*(2)**4 + a3*(2)**3 + a2*(2)**2 + a1*(2) + a0 - 2 = 0
```

```python
eq1_lp = a4*(-2)**4 + a3*(-2)**3 + a2*(-2)**2 + a1*(-2) + a0 - 2
eq2_lp = a4*(-1)**4 + a3*(-1)**3 + a2*(-1)**2 + a1*(-1) + a0 - 1
eq3_lp = a4*(0)**4 + a3*(0)**3 + a2*(0)**2 + a1*(0) + a0 - 0
eq4_lp = a4*(1)**4 + a3*(1)**3 + a2*(1)**2 + a1*(1) + a0 - 1
eq5_lp = a4*(2)**4 + a3*(2)**3 + a2*(2)**2 + a1*(2) + a0 - 2
```

```python
sp.nonlinsolve([eq1_lp, eq2_lp, eq3_lp, eq4_lp, eq5_lp], [a4, a3, a2, a1, a0])
```

```python
# Набор точек, для которого нужно решить задание: (−3, 33), (−2, 31), (−1, 18), (1, −18), (2, −31), (3, −33)
```

```python
# Степень полинома = 5

# f(-3) = 33 => a5 * (-3)**5 + a4 * (-3)**4 + a3 * (-3) ** 3 + a2 * (-3) ** 2 + a1 * (-3) + a0 - 33 = 0
# f(-2) = 31 => a5 * (-2)**5 + a4 * (-2)**4 + a3 * (-2) ** 3 + a2 * (-2) ** 2 + a1 * (-2) + a0 - 31 = 0
# f(-1) = 18 => a5 * (-1)**5 + a4 * (-1)**4 + a3 * (-1) ** 3 + a2 * (-1) ** 2 + a1 * (-1) + a0 - 18 = 0
# f(1) = -18 => a5 * (1)**5 + a4 * (1)**4 + a3 * (1) ** 3 + a2 * (1) ** 2 + a1 * (1) + a0 + 18 = 0
# f(2) = -31 => a5 * (2)**5 + a4 * (2)**4 + a3 * (2) ** 3 + a2 * (2) ** 2 + a1 * (2) + a0 + 31 = 0
# f(3) = -33 => a5 * (3)**5 + a4 * (3)**4 + a3 * (3) ** 3 + a2 * (3) ** 2 + a1 * (3) + a0 + 33 = 0
```

```python
a5, a4, a3, a2, a1, a0 = sp.symbols('a5, a4, a3, a2, a1, a0')
```

```python
eq1_lp = a5 * (-3)**5 + a4 * (-3)**4 + a3 * (-3) ** 3 + a2 * (-3) ** 2 + a1 * (-3) + a0 - 33
eq2_lp = a5 * (-2)**5 + a4 * (-2)**4 + a3 * (-2) ** 3 + a2 * (-2) ** 2 + a1 * (-2) + a0 - 31
eq3_lp = a5 * (-1)**5 + a4 * (-1)**4 + a3 * (-1) ** 3 + a2 * (-1) ** 2 + a1 * (-1) + a0 - 18
eq4_lp = a5 * (1)**5  + a4 * (1)**4  + a3 * (1) ** 3  + a2 * (1) ** 2  + a1 * (1)  + a0 + 18
eq5_lp = a5 * (2)**5  + a4 * (2)**4  + a3 * (2) ** 3  + a2 * (2) ** 2  + a1 * (2)  + a0 + 31
eq6_lp = a5 * (3)**5  + a4 * (3)**4  + a3 * (3) ** 3  + a2 * (3) ** 2  + a1 * (3)  + a0 + 33
```

```python
sp.nonlinsolve([eq1_lp, eq2_lp, eq3_lp, eq4_lp, eq5_lp, eq6_lp], [a5, a4, a3, a2, a1, a0])
```
