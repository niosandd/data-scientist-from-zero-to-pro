```python
from sympy import *
from sympy.plotting import plot3d
x, y = symbols('x, y')
f = x * (y - 5)**2
f
```

```python
fx = diff(f, x)
fx
```

```python
fy = diff(f, y)
fy
```

```python
plot(fx)
```

```python
plot3d(fx)
```

```python
plot3d(fy)
```

```python
solve(fx)
```

```python
solve(fy)
```

```python
a2, a1, a0 = symbols('a2 a1 a0')
```

```python
MSE = 1/3 * (((a2 * 560 + a1 * 1945 + a0) - 4.3)**2 + 
             ((a2 * 340 + a1 * 1495 + a0) - 4.9)**2 + 
             ((a2 * 343 + a1 * 1570 + a0) - 5.2)**2)
MSE
```

```python
MSE2 = diff(MSE, a2)
MSE2
```

```python
MSE1 = diff(MSE, a1)
MSE1
```

```python
MSE0 = diff(MSE, a0)
```

```python
nonlinsolve([MSE2, MSE1, MSE0], [a2, a1, a0])
```
