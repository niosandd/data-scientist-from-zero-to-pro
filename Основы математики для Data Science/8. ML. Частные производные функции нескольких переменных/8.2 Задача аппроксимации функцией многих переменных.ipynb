```python
a2, a1, a0 = symbols('a2, a1, a0')
```

```python
MSE = 1/3 * (((a2 * 2 + a1 * 200 + a0) - 200)**2 + \
             ((a2 * 1 + a1 * 450 + a0) - 300)**2 + \
             ((a2 * 3 + a1 * 550 + a0) - 600)**2)
MSE
```

```python
{a2: 1, a1: 2, a0: 3}
```

```python
# f1 = 100 * x2 - 1 * x1 - 100
# f2 = 200 * x2 + 3 * x1 + 20
```

```python
MSE.subs({a2: 100, a1: -1, a0: -100})
```

```python
MSE.subs({a2: 200, a1: 3, a0: 20})
```

```python
# 3 точки => 2 полинома
# 10, 30, 7
# -5, 15, 20
# 16, 31, -4

MSE = (((a2 * 30 + a1*10 + a0) - 7)**2 + \
       ((a2 * 15 + a1*(-5) + a0) - 20)**2 + \
       ((a2 * 31 + a1*16 + a0) - (-4)) **2) * 1/2

MSE
```

```python
MSE.subs({a2: -2, a1: 1, a0: -7})
```

```python
MSE.subs({a2: 20, a1: 3, a0: -4})
```
