```python
from sympy import *
import numpy
```

```python
x, y, z = symbols('x, y, z')

MSE = 1 / 6 * ((5 * x + 7 * y - 5 * z + 47)**2 + \
               (- 2 * y + 2 * z - 10)**2 + \
               (-4 * x - 8 * y - 7 * z - 63)**2 + \
               (x + y + 2 * z + 1)**2 + \
               (2 * x - y + 2 * z + 4)**2 + \
               (4 * x + y + 4 * z + 2)**2)
MSE.subs({x: 0, y: 0, z: 0})
```

```python
MSEx = diff(MSE, x)
MSEx
```

```python
MSEy = diff(MSE, y)
MSEy
```

```python
MSEz = diff(MSE, z)
MSEz
```
