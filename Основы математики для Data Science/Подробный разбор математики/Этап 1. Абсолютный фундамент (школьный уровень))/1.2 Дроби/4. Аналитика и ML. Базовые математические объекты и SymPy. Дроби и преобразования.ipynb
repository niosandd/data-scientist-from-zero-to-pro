```python
from sympy import *
```

Дробь $\dfrac{m}{n}$ есть частное $m:n$. Знаменатель $n$ — натуральное, на сколько равных частей разрезали целое. Числитель $m$ — целое, сколько частей взяли. Черта и есть деление, поэтому $6:11=\dfrac{6}{11}$. Такие числа образуют $\mathbb{Q}$. Целое входит туда как дробь со знаменателем $1$: $5=\dfrac{5}{1}$.

`S` переводит число в тип SymPy, чтобы деление не стало float.

```python
S(6)/11
S(-8)/8
S(12)/1
```

Если $|m|<n$, дробь правильная. Если $|m|\ge n$ — неправильная. Смешанное число — целое плюс правильная дробь: $1\dfrac{20}{60}=1+\dfrac{20}{60}$.

Неправильная → смешанная: деление с остатком. $80:60=1$, остаток $20$, значит $\dfrac{80}{60}=1\dfrac{20}{60}$.

Смешанная → неправильная: целую часть умножают на знаменатель и прибавляют числитель.

$$
1\dfrac{20}{60}=\dfrac{1\cdot 60+20}{60}=\dfrac{4}{3}
$$

`divmod` как раз снимает частное и остаток. `Rational` сразу сокращает.

```python
Rational(40, 60)
Rational(80, 60)
divmod(80, 60)
(1*60 + 20)/S(60)
```

Основное свойство: числитель и знаменатель можно умножить или разделить на одно и то же число, не равное нулю. Значение не изменится.

$$
\frac{a}{b}=\frac{an}{bn}=\frac{a:n}{b:n}
$$

Сокращение — деление на общий делитель. Самый большой такой делитель — НОД. Если он равен $1$, дробь несократима.

Для $63$ и $84$ алгоритм Евклида: большее делят на меньшее, затем делитель на остаток, пока остаток не станет нулём. Последний ненулевой остаток и есть НОД.

$$
84=63\cdot 1+21,\qquad 63=21\cdot 3+0 \implies \mathrm{НОД}(63,84)=21
$$

Через простые множители берут общие простые в меньшей степени: $63=3^{2}\cdot 7$, $84=2^{2}\cdot 3\cdot 7$, общие $3$ и $7$, произведение $21$. Тогда

$$
\frac{63}{84}=\frac{63:21}{84:21}=\frac{3}{4}
$$

Нечётное число минусов делает дробь отрицательной, чётное — положительной.

`gcd` считает Евклида. `factorint` раскладывает на простые. `Rational(63, 84)` сам делит на этот НОД.

```python
gcd(63, 84)
factorint(63)
factorint(84)
Rational(63, 84)
Rational(-3, 5)
Rational(-3, -5)
```

Разные знаменатели приводят к общему. Берут НОК — наименьшее число, которое делится на каждый знаменатель.

Кратные $3$: $3,6,9,12,15,18,21,24,\ldots$. Кратные $8$: $8,16,24,\ldots$. Первое общее — $24$.

Через множители: $3=3$, $8=2^{3}$. В НОК каждый простой в наибольшей степени: $2^{3}\cdot 3=24$.

Через НОД:

$$
\mathrm{НОК}(a,b)=\frac{ab}{\mathrm{НОД}(a,b)}=\frac{3\cdot 8}{1}=24
$$

Дополнительный множитель — частное общего знаменателя и старого: $24:3=8$, $24:8=3$. Поэтому

$$
\frac{2}{3}=\frac{16}{24},\qquad \frac{5}{8}=\frac{15}{24},\qquad 16>15\implies\frac{2}{3}>\frac{5}{8}
$$

С одинаковыми знаменателями работают только с числителями:

$$
\frac{a}{n}\pm\frac{b}{n}=\frac{a\pm b}{n},\qquad
\frac{1}{6}+\frac{3}{8}=\frac{4}{24}+\frac{9}{24}=\frac{13}{24}
$$

`ilcm` даёт общий знаменатель. Операции `+` и `>` у `Rational` считают значение; `ilcm` нужен, когда сам знаменатель надо увидеть.

```python
ilcm(3, 8)
3 * 8 // gcd(3, 8)
Rational(2, 3) > Rational(5, 8)
Rational(1, 6) + Rational(3, 8)
Rational(9, 4) + Rational(5, 3)
```

Произведение — числитель на числитель, знаменатель на знаменатель, затем сокращают. Предлог «от» означает умножение: $\dfrac{2}{3}$ от $48$ это $\dfrac{2}{3}\cdot 48=32$.

Деление — умножение на обратную. Дроби $\dfrac{a}{b}$ и $\dfrac{b}{a}$ в произведении дают $1$.

$$
\frac{2}{3}:\frac{5}{7}=\frac{2}{3}\cdot\frac{7}{5}=\frac{14}{15}
$$

Если $\dfrac{2}{3}$ числа равны $48$, само число равно $48:\dfrac{2}{3}=72$.

Оператор `/` у двух `Rational` сам умножает на обратную. Явно обратную показывает `1 / дробь`.

```python
Rational(6, 35) * Rational(14, 9)
Rational(2, 3) * 48
Rational(2, 3) / Rational(5, 7)
1 / Rational(3, 8)
48 / Rational(2, 3)
```

Десятичная дробь — обыкновенная со знаменателем $10^{n}$: $0.25=\dfrac{25}{100}=\dfrac{1}{4}$. Строку `"0.25"` передают в `Rational`, потому что голый float $0.25$ в двоичной записи может дать не ту дробь.

Умножение и деление на $10^{n}$ сдвигает точку. Делитель делают целым, умножив оба числа на одну и ту же степень десяти: $12.5:0.5$ после умножения на $10$ есть $125:5=25$. В коде этот сдвиг уже внутри деления двух `Rational`.

`S(10)**n` держит степень десятки точной. Обычный `10**(-3)` в Python уже float.

```python
Rational("0.25")
Rational("12.3") - Rational("4.075")
Rational("3.14") * S(10)**2
Rational("12.5") / Rational("0.5")
```

Конечная десятичная получается, когда после сокращения в знаменателе остаются только $2$ и $5$. Иначе период. Период переводят уравнением. Для $0.(3)$: $x=0.(3)$, $10x=3.(3)$, вычитают, $9x=3$, $x=\dfrac{1}{3}$. Для $0.1(6)$: $10x=1.(6)$, $100x=16.(6)$, $90x=15$, $x=\dfrac{1}{6}$.

`symbols` заводит неизвестную. `Eq` оставляет равенство невычисленным: обычный `==` сразу дал бы True или False. `solve` находит корень, `[0]` берёт его из списка. `N` только показывает десятичный вид.

```python
N(S(1)/7, 20)
x = symbols("x")
solve(Eq(10*x - x, 3), x)[0]
solve(Eq(100*x - 10*x, 15), x)[0]
```

Отрицательная степень: $a^{-n}=\dfrac{1}{a^{n}}$. Показатели при одном основании складывают при умножении и вычитают при делении.

$$
\frac{2^{-3}\cdot 5^{2}}{2^{4}\cdot 5^{-1}}=2^{-3-4}\cdot 5^{2-(-1)}=2^{-7}\cdot 5^{3}=\frac{125}{128}
$$

Стандартный вид: $a\cdot 10^{n}$ при $1\le|a|<10$. Слагаемые с разными порядками приводят к одной степени десятки. Единицы переводят умножением или делением на связь между ними; м/с в км/ч умножают на $3{,}6$.

`simplify` собирает степени одного основания в короткий ответ.

```python
S(10)**(-3)
simplify((S(2)**(-3)*S(5)**2) / (S(2)**4 * S(5)**(-1)))
Rational("2.1")*S(10)**5 + Rational("3.4")*S(10)**4
18 * Rational("3.6")
72 / Rational("3.6")
```

Порядок действий везде один: скобки, степени, умножение и деление слева направо, сложение и вычитание слева направо.

Задание 1. Квадрат суммы $(u+v)^{2}=u^{2}+2uv+v^{2}$. Скобку раскрывают распределительным законом.

$$
(2x+3y)^{2}-3x\Bigl(\frac{4x}{3}+4y\Bigr)
=4x^{2}+12xy+9y^{2}-4x^{2}-12xy
=9y^{2}
$$

При $y=\sqrt{3}$ получается $9\cdot 3=27$. Икс в ответ не входит, поэтому $1{,}038$ его не меняет.

`symbols` вводит буквы. `simplify` сворачивает многочлен. `subs` подставляет. `sqrt` — точный корень.

```python
x, y = symbols("x y")
expr = (2*x + 3*y)**2 - 3*x * (4*x/3 + 4*y)
expr
simplify(expr)
simplify(expr).subs({x: 1.038, y: sqrt(3)})
```

Задание 2. Квадрат разности $(u-v)^{2}=u^{2}-2uv+v^{2}$. Среднее слагаемое уничтожает $28ab$.

$$
28ab+(2a-7b)^{2}=28ab+4a^{2}-28ab+49b^{2}=4a^{2}+49b^{2}
$$

При $a=\sqrt{15}$, $b=\sqrt{8}$:

$$
4\cdot 15+49\cdot 8=60+392=452
$$

```python
a, b = symbols("a b")
expr = 28*a*b + (2*a - 7*b)**2
simplify(expr)
simplify(expr).subs({a: sqrt(15), b: sqrt(8)})
```

Задание 3. Сначала доли с общим знаменателем $10$, затем сокращают одну $a$.

$$
\Bigl(\frac{a}{2}+\frac{a}{5}\Bigr)\cdot\frac{1}{a^{2}}
=\frac{1}{a}\Bigl(\frac{5}{10}+\frac{2}{10}\Bigr)
=\frac{7}{10a}
$$

При $a=0{,}7$:

$$
\frac{7}{10\cdot 0{,}7}=1
$$

Для подстановки берут `S("0.7")` или `Rational(7, 10)`, чтобы не получить хвост от двоичного `0.7`.

```python
a = symbols("a")
expr = (a/2 + a/5) * 1/a**2
simplify(expr)
simplify(expr).subs({a: S("0.7")})
```

Задание 4. Дробь над дробью — умножение на обратную. Сначала числитель к знаменателю $x+2$.

$$
x-\frac{6x}{x+2}=\frac{x(x+2)-6x}{x+2}=\frac{x^{2}-4x}{x+2}
$$

$$
\frac{\dfrac{x^{2}-4x}{x+2}}{\dfrac{x-4}{x+2}}
=\frac{x(x-4)}{x+2}\cdot\frac{x+2}{x-4}
=x
$$

при $x\neq -2$ и $x\neq 4$. В точке $x=5$ значение $5$.

```python
x = symbols("x")
expr = (x - 6*x/(x+2)) / ((x-4)/(x+2))
simplify(expr)
simplify(expr).subs({x: 5})
```

Задание 6. Запись `5/c/k**2` читается слева направо как $\dfrac{5}{c k^{2}}$. Затем умножение на $k^{3}$ и на $c$ даёт $5k$.

$$
5k-\frac{10k+1}{2}=\frac{10k-10k-1}{2}=-\frac{1}{2}
$$

Буквы сокращаются полностью.

```python
k, c = symbols("k c")
expr = 5/c/k**2 * k**3 * c - (10*k + 1)/2
simplify(expr)
```

Сложная часть. Знаменатель первой дроби $x^{2}-2x=x(x-2)$. Сокращают $x$ при $x\neq 0$:

$$
\frac{3x^{2}+4x}{x(x-2)}=\frac{3x+4}{x-2}
$$

Две дроби с знаменателем $x-2$ объединяют:

$$
\frac{3x+4}{x-2}-\frac{x+8}{x-2}=\frac{2x-4}{x-2}=2
$$

Остаётся

$$
2-\frac{2x+7}{x}=2-2-\frac{7}{x}=-\frac{7}{x}
$$

при $x\neq 0$ и $x\neq 2$. `simplify` проходит эту цепочку сам.

```python
x = symbols("x")
expr = (3*x**2 + 4*x)/(x**2 - 2*x) - (2*x + 7)/x - (x + 8)/(x - 2)
expr
simplify(expr)
```
