* **Натуральные числа** - это те числа, которые используются при счёт объектов (множество всех натуральных чисел обозначают **ℕ**). Важнейшая функция натуральных чисел — обозначать количество объектов. Также это не только подсчет. Обычное число это тоже натурально число. Важно понимать, что числа отрицательные, дробные и любые другие, это не натуральное.
* **Число** - это обычная запись цифрами (например: 2, 17, 148)
* **Текст числа** - "два", "семнадцать", "сто сорок восемь" (программисты назвали бы это текстом, лингвисты - числительным)

*Краткая выжимка*
