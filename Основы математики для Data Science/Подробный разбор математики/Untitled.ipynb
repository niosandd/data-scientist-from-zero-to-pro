# Натуральные числа и основы арифметики

## Кратко

- **Натуральные числа** — это числа, которые используются при счёте объектов  
  (множество всех натуральных чисел обозначают **ℕ**).  
  Важнейшая функция натуральных чисел — обозначать **количество** объектов.  
  Обычное положительное целое число также является натуральным.  
  Отрицательные, дробные и любые другие числа **не являются** натуральными.

- **Число** — запись цифрами  
  *Пример:* `2`, `17`, `148`

- **Текст числа** — словесная запись  
  *Пример:* «два», «семнадцать», «сто сорок восемь»  
  (программисты называют это текстом, лингвисты — числительным)

- Натуральные числа имеют две теории:
  - **количественная**
  - **порядковая**

- **Ноль** является **целым** числом (не натуральным).

## Разряды и классы

**Число:** `456 654 667`

- В числе **667**:
  - разряд единиц — **7**
  - разряд десятков — **6**
  - разряд сотен — **6**

**Классы чисел:**  
единица → десяток → сотня → тысяча → миллион → миллиард → …

## Чётные числа

**Чётные числа** — те, которые делятся на 2 без остатка (остаток = 0).

---

## Компоненты действий

### Сложение
m + n = k
- слагаемое + слагаемое = **сумма**

### Вычитание
k − n = m
- уменьшаемое − вычитаемое = **разность**

### Свойства нуля
* k + 0 = k
* k − 0 = k

При прибавлении или вычитании нуля число **не меняется**.

---

## Свойства операций

### Коммутативность (переместительный закон)
n + m = m + n

### Некоммутативность вычитания
n − m ≠ m − n

## Выражения

| Тип                  | Пример              |
|----------------------|---------------------|
| **Числовое выражение**   | `5 + 6 = 11`       |
| **Буквенное выражение**  | `a + b` `5a + 3c`  |

Если выполнить все действия в числовом выражении, получают **значение выражения**.
