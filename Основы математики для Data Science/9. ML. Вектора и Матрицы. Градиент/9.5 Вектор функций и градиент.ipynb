```python
from sympy import *
from sympy.plotting import plot3d
import numpy as np
```

```python
# Первая итерация
```

```python
a1, a2 = symbols('a1 a2')
MSE = 1 / 4 * ((a1 * 1 + a2 * 2 - 5)**2 + \
               (a1 * 5 + a2 * 3 - 6)**2 + \
               (a1 * 2 + a2 * 4 - 10)**2 + \
               (a1 * 3 + a2 * 7 - 8)**2)

print(MSE)
```

```python
MSE1 = diff(MSE, a1)
MSE2 = diff(MSE, a2)
print(MSE1)
print(MSE2)
```
