```python
from sympy import *
from sympy.plotting import plot3d
```

```python
init_printing(use_unicode=False, wrap_line=False, no_global=True)
```

```python
x,y,z = symbols("x, y, z")
```

```python
f = 0.8 * x + 20 * y
f
```

```python
plot3d(f)
```

```python
f = x*y / sqrt(x**2 + y**2 -5)
f
```

```python
plot3d(f)
```

```python
f = sin(sqrt(x**2+y**2))
f
```

```python
plot3d(f)
```

```python
# f = E ** (0.9*x**2 - 0.45 * (x-y)**2)
f = exp(-0.9 * (x**2 + 0.5 * (x - y)**2))
```

```python
plot3d(f, (x, -3, 3), (y, -3, 3))
```

```python
f = x * cos(y)
```

```python
plot3d(f)
```
