```python
from sympy import *
from sympy.plotting import plot3d
```

```python
init_printing(use_unicode=False, wrap_line=False, no_global=True)
```

```python
x,y = symbols('x, y')
```

```python
f = 5 * x - 3 * y + 2
f
```

```python
solve(f)
```

```python
func_x = 3 / 5 * y - 2 / 5
func_x
```

```python
plot(func_x)
```

```python
plot3d(f)
```
