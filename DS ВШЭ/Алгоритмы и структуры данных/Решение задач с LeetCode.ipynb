```python
nums = [1, 1, 2, 2, 3, 4, 4]
n = len(nums)
index = 0

while index < n - 1:
    if nums[index] != nums[index + 1]:
        print(nums[index])
        break
    index += 2
else:
    print(nums[-1])
```

```python
Дан массив nums размера n, вернуть элемент большинства.

Элемент большинства — это элемент, который встречается в массиве более ⌊n / 2⌋ раз.Можно предположить, что элемент большинства всегда присутствует в массиве.

Входные данные: nums = [3,2,3]
Выходные данные: 3
```

```python
nums = [2, 2, 2, 3,3,3]
n = len(nums)
max_number = 0
```

```python
for num in nums:
    if nums.count(num) > n //2:
        if nums.count(num) > max_number:
            max_number = nums.count(num)
        print(num)
```
