```python
a = int(input())
b = int(input())
c = int(input())
k = min(a, c, b // 2)

ans = 4 * k
a -= k
b -= 2 * k
c -= k

if a > 0:
    ans += 1
    if b > 0:
        ans += 1
        if c > 0:
            ans += 1
            if b > 1:
                ans += 1

print(ans)
```
